```python
carts_df["calculated_total_products"] = carts_df["products"].apply(len)

carts_df["calculated_total_quantity"] = carts_df["products"].apply(
    lambda products: sum(product["quantity"] for product in products)
)

print(
    "Diferencias en totalProducts:",
    (carts_df["totalProducts"] != carts_df["calculated_total_products"]).sum()
)

print(
    "Diferencias en totalQuantity:",
    (carts_df["totalQuantity"] != carts_df["calculated_total_quantity"]).sum()
)
```

```python
carts_df["calculated_total"] = carts_df["products"].apply(
    lambda products: sum(product["total"] for product in products)
)

carts_df["calculated_discounted_total"] = carts_df["products"].apply(
    lambda products: sum(product["discountedTotal"] for product in products)
)

print(
    "Diferencias en total:",
    (~np.isclose(carts_df["total"], carts_df["calculated_total"], atol=0.01)).sum()
)

print(
    "Diferencias en discountedTotal:",
    (~np.isclose(carts_df["discountedTotal"], carts_df["calculated_discounted_total"], atol=0.01)).sum()
)
```

```python
# Usuarios referenciados por los carritos
cart_user_ids = set(carts_df["userId"])
user_ids = set(users_df["id"])

missing_users = cart_user_ids - user_ids

# Productos referenciados por los carritos
cart_product_ids = {
    product["id"]
    for products in carts_df["products"]
    for product in products
}

product_ids = set(products_df["id"])

missing_products = cart_product_ids - product_ids

print("Usuarios inexistentes:", len(missing_users))
print("Productos inexistentes:", len(missing_products))
```
